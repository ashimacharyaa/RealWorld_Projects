```python
#defining custom function
def financial_report(company_name:str, revenue:int, expenses:int) -> str:
    """ Generate financial report of a company that calculate net income"""
    net_income = revenue-expenses

    report = f"Financial report for {company_name}\n"
    report += f"Revenue:{revenue}\n"
    report += f"Expenses:{expenses}\n"
    report += f"Net Income:{net_income}\n"

    return report
```

```python
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash",api_key="--------Q") 
tools = load_tools(["llm-math"], llm=llm)


@tool
def financial_report(company_name:str, revenue:float, expenses:float) -> str:
    """ Generate financial report for a company through netprofit"""
    net_income = revenue - expenses

    report = f"Financial report fo {company_name}:\n"
    report += f"Revenue:{revenue}\n"
    report +=f"Expenses:{expenses}\n"
    report +=f"Net Profit:{net_income}\n"
    return report
```

```python
result = financial_report.invoke({
    "company_name": "Ashim AI",
    "revenue": 50000,
    "expenses": 35000
})

print(result)
```

```python
from langgraph.prebuilt import create_react_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import os

# Initialize Gemini
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key="AQ.Ab8RN6J-nehL1yOG9fSkcmJTee5FVpIFoThbzdYViViSd_HrTQ"
)

# Create agent with custom tool
agent = create_react_agent(
    llm,
    [financial_report]
)

# Invoke agent
response = agent.invoke({
    "messages": [
        ("human", "TechStack generated $10 million in revenue "
                  "with $8 million in costs. Generate a financial report.")
    ]
})

# Print the final AI response
print(response["messages"][-1].content)
```
