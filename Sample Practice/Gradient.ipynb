```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    'Size':[1000,1200,1500,1800,2000],
    'Price':[200,240,300,360,400]
})
```

```python
print(df)
```

```python
x = df['Size'].values
y = df['Price'].values
```

```python
print(x,y)
```

```python
#feature scaling
x_mean = x.mean() 
x_std = x.std()
```

```python
x_scaled = (x-x_mean)/x_std
```

```python
print("\n Scaled X:")
print(x_scaled)
```

```python
w=0
b=0
```

```python
learning_rate = 0.01
epochs = 10000
n=len(x)
```

```python
#gradient descent
cost_history = []

for epoch in range(epochs):
    y_pred = w*x_scaled+b
```

```python
#error
error = y_pred - y
```

```python
#cost/mse
cost = np.mean(error ** 2)
```

```python
#gradients
dw = (-2/n)*np.sum(x_scaled*error)
db = (-2/n)*np.sum(error)
```

```python
#update parameters
w = w-learning_rate*dw
b = b-learning_rate*db
```

```python
cost_history.append(cost)
```

```python
print("Final_Weight",w)
print("Final Bias",b)
print("Final Cost",cost_history[-1])
```

```python
#prediction
y_pred = w*x_scaled+b
```

```python
print("\n Actual vs Predicted")

result = pd.DataFrame({
    'Actual':y,
    'Prediction':y_pred
})

print(result)
```

```python
plt.plot(cost_history)

plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.title("Gradient Descent - Cost Reduction")

plt.show()
```
