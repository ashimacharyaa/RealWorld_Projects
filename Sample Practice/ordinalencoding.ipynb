```python
from sklearn.preprocessing import OrdinalEncoder
import numpy as np
import pandas as pd
```

```python
data = {
    "education": ["Under","Grad","PHD"],
    "salary": ["Low","Medium","High"]
}
```

```python
df = pd.DataFrame(data)

print(df)
```

```python
encoder = OrdinalEncoder(
    categories = [
        ["Under","Grad","PHD"],
        ["Low","Medium","High"]
    ]
)
```

```python
df_encoding = encoder.fit_transform(df)
                                        
df_encoding = pd.DataFrame(
    df_encoding,
    columns = df.columns
)
```

```python
print(df_encoding)
```
