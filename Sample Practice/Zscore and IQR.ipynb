```python
import pandas as pd
from scipy.stats import zscore
```

```python
df = pd.DataFrame({
    'Age':[21,22,23,24,26,90]
})
```

```python
df['z_score'] = zscore(df['Age'])
```

```python
print(df)
```

```python
df_clean = df[abs(df['z_score']) <= 3]

print(df_clean)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Salary': [20,22,21,23,25,24,26,27,100]
})
```

```python
q1 = df['Salary'].quantile(0.25)
q3 = df['Salary'].quantile(0.75)
```

```python
IQR = q3 - q1
```

```python
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR
```

```python
print("q1 =",q1)
print("q3 =",q3)
```

```python
print("Lower Bound:", lower_bound)
print("Upper_Bound:",upper_bound)
```

```python
df_cleaned = df[
    (df['Salary'] >= lower_bound) &
    (df['Salary'] <= upper_bound)
]
```

```python
print("Original Data:",df)
print("Cleaned Data:",df_cleaned)
```
