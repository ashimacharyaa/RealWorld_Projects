```python
import pandas as pd
from sklearn.preprocessing import FunctionTransformer
```

```python
df = pd.DataFrame({
    'Age':[10,20,30,40,50]
})

print(df)
```

```python
print(df)
```

```python
def square(x):
    return x ** 2
```

```python
transformer = FunctionTransformer(square)
```

```python
X_transformed = transformer.fit_transform(df)
print(X_transformed)
```

```python
#Log Transformation
import numpy as np
from sklearn.preprocessing import FunctionTransformer
```

```python
transformer = FunctionTransformer(np.log1p)
```

```python
X_transformed = transformer.fit_transform(df)

print(X_transformed)
```

```python
def add_10(x):
    return x + 10

transformer = FunctionTransformer(add_10)

X_transformed = transformer.fit_transform(df)

print(X_transformed)
```
