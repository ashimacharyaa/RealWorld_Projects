```python

class University:

    # Class variables
    university_name = "Global University"
    total_students = 0
    passing_marks = 40

    # Constructor
    def __init__(self, student_name: str, student_id: str,
                 department: str, fees: float = 50000):

        self.student_name = student_name
        self.student_id = student_id
        self.department = department
        self.fees = fees

        self.subjects = {}
        self.grades = {}

        University.total_students += 1

    # Enroll subject
    def enroll_subject(self, subject: str, credit: int):

        if subject in self.subjects:
            return "Already Enrolled in This Subject"

        elif credit <= 0:
            return "Invalid Credit"

        elif len(self.subjects) >= 6:
            return "Maximum Subject Limit Reached"

        else:
            self.subjects[subject] = credit
            return f"{subject} Enrolled Successfully"

    # Add grades
    def add_grade(self, subject: str, marks: float):

        if subject not in self.subjects:
            return "You Are Not Enrolled in This Subject"

        elif not 0 <= marks <= 100:
            return "Invalid Marks"

        elif subject in self.grades:
            return "Grade Already Added"

        else:
            self.grades[subject] = marks
            return f"Grade Added Successfully for {subject}"

    # Calculate GPA
    def calculate_gpa(self):

        if not self.grades:
            return "No grades available"

        total_gpa = 0

        for subject, marks in self.grades.items():

            if marks >= 90:
                gpa = 4.0

            elif marks >= 80:
                gpa = 3.6

            elif marks >= 70:
                gpa = 3.2

            elif marks >= 60:
                gpa = 2.8

            elif marks >= 50:
                gpa = 2.4

            elif marks >= 40:
                gpa = 2.0

            else:
                gpa = 0.0

            total_gpa += gpa

        return round(total_gpa / len(self.grades), 2)

    # Calculate total fees
    def calculate_total_fees(self):

        total_credits = sum(self.subjects.values())

        total_fees = self.fees + (total_credits * 1500)

        return total_fees

    # Display student information
    def display_info(self):

        print("\n========== STUDENT INFORMATION ==========")

        print("University:", University.university_name)
        print("Student Name:", self.student_name)
        print("Student ID:", self.student_id)
        print("Department:", self.department)

        print("\nEnrolled Subjects:")

        for subject, credit in self.subjects.items():
            print(f"{subject}: {credit} credits")

        print("\nGrades:")

        for subject, marks in self.grades.items():
            print(f"{subject}: {marks}")

        print("\nGPA:", self.calculate_gpa())
        print("Total Fees:", self.calculate_total_fees())

        print("=========================================")


# Creating student objects

ashim = University("Ashim", "ST001", "Computer Science")

ram = University("Ram", "ST002", "Data Science", 60000)


# Subject enrollment

print(ashim.enroll_subject("Python", 3))
print(ashim.enroll_subject("Statistics", 4))
print(ashim.enroll_subject("Machine Learning", 3))

print(ram.enroll_subject("Python", 3))
print(ram.enroll_subject("Mathematics", 4))


# Adding grades

print(ashim.add_grade("Python", 95))
print(ashim.add_grade("Statistics", 85))
print(ashim.add_grade("Machine Learning", 75))

print(ram.add_grade("Python", 80))
print(ram.add_grade("Mathematics", 90))


# GPA calculation

print("\nAshim GPA:", ashim.calculate_gpa())
print("Ram GPA:", ram.calculate_gpa())


# Fee calculation

print("Ashim Fees:", ashim.calculate_total_fees())
print("Ram Fees:", ram.calculate_total_fees())


# Display information

ashim.display_info()
ram.display_info()


# Total students

print("\nTotal Students:", University.total_students)
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using credit card")

class Esewa(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using Esewa")

class Khalti(Payment):

    def pay(self, amount):
        print(f"Paid Rs {amount} using Khalti")
```

```python
payment2 = Khalti()
payment2.pay(50000)
```

```python
from abc import ABC, abstractmethod

class Taking(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(Taking):

    def pay(self, amount):
        print(f"Took Rs. {amount} using credit card")

class IME_PAY(Taking):
    def pay(self, amount):
        print(f"Paid using {amount} through IME PAY")

class Khalti(Taking):
    def pay(self, amount):
        print(f"Paid {amount} using Khalti")
```
