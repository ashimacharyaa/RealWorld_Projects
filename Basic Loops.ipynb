```python
numbers = [10, 30, 45, 30, 10, 20]

duplicate = []
unique = []

for num in numbers:
    if num in unique:
        if num not in duplicate:
            duplicate.append(num)
    else:
        unique.append(num)

print("Duplicates:", duplicate)
print("Unique:", unique)
```

```python
numbers = [10, -5, 20, -8, 0, 15, -2]
pos=[]
neg=[]
zero=[]
for num in numbers:
    if num > 0:
        pos.append(num)
    elif num == 0:
        zero.append(num)
    else:
        neg.append(num)

print(f"Positive Numbers:{pos}")
print(f"Negative Numbers:{neg}")
print(f"Zero Numbers:{zero}")
```

```python
numbers = [2, 3, 2, 4, 3, 2, 5, 4, 3]
frequency={}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

for key, value in frequency.items():
    print(f"{key} appears {value} times")
```

```python
numbers = [45, 12, 89, 3, 67, 23]

largest=[0]
smallest=[0]

for num in numbers:
    if num > largest:
        largest=num

    if num < smallest:
        smallest=num

print("Largest",largest)
print("Smallest",smallest)
```
