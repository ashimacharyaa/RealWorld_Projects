```python
#tokenizer
from nltk.tokenize import word_tokenize, sent_tokenize
text = "I am Ashim Acharya. I am gentleman and i am profoundly proud of myself to have such an amazing friends"

cn = word_tokenize(text)
print(cn)
```

```python
#split_sentences
sentences = sent_tokenize(text)
print(sentences)
```

```python
#stopwords
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
text = "I love learning machine learning"
words = word_tokenize(text)

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

```python
#stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing","played","plays","player"]

for word in words:
    print(word, "-", stemmer.stem(word))
```

```python
#lemmatization
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running","cars","better"]

for word in words:
    print(word, "-", lemmatizer.lemmatize(word))
```
